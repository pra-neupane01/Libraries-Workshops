```python
import cv2
import numpy as np
image1 = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Images\binary image 1.png")
image2 = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Images\binary image 2.png")

re_img1 = cv2.resize(image1, (250,250))
re_img2 = cv2.resize(image2, (250,250))

# new_img = cv2.bitwise_and(re_img1, re_img2)
# new_img = cv2.bitwise_or(re_img1, re_img2)
new_img = cv2.bitwise_xor(re_img1, re_img2)
# new_img = cv2.bitwise_not(re_img1, re_img2)

h = np.hstack((re_img1, re_img2, new_img))

cv2.imshow("window",h)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

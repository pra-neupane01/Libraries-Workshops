```python
import cv2
image = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Resources\sharbhari.jpg")

reduced1 = cv2.pyrDown(image)
reduced2 = cv2.pyrDown(reduced1)

increased1 = cv2.pyrDown(image)
increased2 = cv2.pyrUp(increased1)

cv2.imshow("window0",image)
cv2.imshow("window1",reduced1)
cv2.imshow("window2",reduced2)
cv2.imshow("window3",increased1)
cv2.imshow("window4",increased2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

image = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Images\sharbhari.jpg")
image = cv2.resize(image, (400,500))

g = cv2.GaussianBlur(image,(5,5),0) # kernel size (5,5) sigmax = 0  | general smooting before edge detection
m = cv2.medianBlur(image, 5) # kernel size 5   #smooth skin without blurring edge
b = cv2.bilateralFilter(image,10,100,100)  # d= diameter of each pixel neighbourhood
h = np.hstack((image,g,m,b))
cv2.imwrite(h,r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Images\blurred_image.png")
cv2.imshow("window",h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

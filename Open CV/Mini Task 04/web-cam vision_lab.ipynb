```python
import cv2
import time

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("unable to open camera")
    exit()

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = 30

# Setting container
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter("webcam_output.mp4", fourcc, fps, (frame_width, frame_height))

prev_time = time.time()
recording = False
shot_count = 0

print("Press S -> save screenshot")
print("Press R -> toggle recording")
print("Press Q -> quit")

while True:
    success, frame = cap.read()
    if not success:
        print("could not grab frame")
        break

    # 1. Mirror image
    frame = cv2.flip(frame, 1)

    # 2. Display FPS & Latency
    current_time = time.time()
    time_diff = current_time - prev_time
    actual_fps = int(1 / time_diff) if time_diff > 0 else 0
    prev_time = current_time
    cv2.putText(frame, f"FPS: {actual_fps}", (20, 40), cv2.FONT_HERSHEY_COMPLEX, 0.8, (0, 255, 0), 2)

    # 3. Show Resolution
    cv2.putText(frame, f"Res: {frame_width}x{frame_height}", (20, 80), cv2.FONT_HERSHEY_COMPLEX, 0.8, (0, 255, 0), 2)

    # 4. Record if mode active
    if recording:
        out.write(frame)
        cv2.putText(frame, "REC", (frame_width - 80, 40), cv2.FONT_HERSHEY_COMPLEX, 0.8, (0, 0, 255), 2)

    # 5. Display Window
    cv2.imshow("Real-Time Computer Vision", frame)

    # 6. Key Controls
    key = cv2.waitKey(1) & 0xFF

    if key == ord("s") or key == ord("S"):
        shot_count += 1
        cv2.imwrite(f"screenshot_{shot_count}.png", frame)
        print(f"Screenshot saved: screenshot_{shot_count}.png")

    elif key == ord("r") or key == ord("R"):
        recording = not recording
        status = "Started" if recording else "Paused"
        print(f"Recording {status}")

    elif key == ord("q") or key == ord("Q"):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
image = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Resources\flower bit.png")
re_image = cv2.resize(image,(400,450))


 
opening_image = cv2.morphologyEx(re_image,
                                cv2.MORPH_OPEN,
                                kernel= np.array((10,10), np.int8),
                                iterations = 2)

closing_image = cv2.morphologyEx(re_image,
                                cv2.MORPH_CLOSE,
                                kernel= np.array((10,10), np.int8),
                                iterations = 2)

gradient_image = cv2.morphologyEx(re_image,
                                cv2.MORPH_GRADIENT,
                                kernel= np.array((10,10), np.int8),
                                iterations = 2)

cv2.imshow("window1", re_image)
cv2.imshow("window2", opening_image)
cv2.imshow("window3", closing_image)
cv2.imshow("window3", gradient_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2 
import numpy as np

img = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Images\sharbhari.jpg")
re_img = cv2.resize(img, (380, 500))

new_img = cv2.polylines(
    re_img,
    [np.array([[110,110],[220,60],[300,100],[300,250],[100,250]])],
    True,                   # closed or not  
    (0,255,0),              # color
    4,
    16
)

cv2.imshow("My window", new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

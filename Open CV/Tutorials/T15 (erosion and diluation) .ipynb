```python
import cv2
import numpy as np
image = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Resources\flower bit.png")
re_image = cv2.resize(image,(400,450))
 
eroded_image = cv2.erode(re_image,
          kernel= np.array((10,10), np.int8),
          iterations= 1)

diluted_image = cv2.dilate(re_image,
                           kernel= np.array((10,10), np.int8),
                           iterations= 1)

cv2.imshow("window1", re_image)
cv2.imshow("window2", eroded_image)
cv2.imshow("window3", diluted_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

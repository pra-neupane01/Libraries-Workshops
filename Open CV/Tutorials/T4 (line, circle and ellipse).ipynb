```python
import cv2

img = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Images\SRK.jpg")
re_img = cv2.resize(img, (600, 400))

new_img = cv2.ellipse(
    re_img,                 # img
    (310, 110),             # center
    (55, 100),              # axes (x-radius, y-radius)
    0 ,                     # angle
    0,                      # startAngle
    360,                    # endAngle
    (255, 0, 0),            # color (BGR)
    2,                      # thickness
    cv2.LINE_4              # lineType
)

cv2.imshow("My window", new_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Open default webcam (0)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not access the webcam.")
    exit()

# Default mode set to original (6)
mode = ord("6")

print("--- Webcam Processing Controls ---")
print("Press 1 -> Grayscale")
print("Press 2 -> HSV")
print("Press 3 -> Gaussian Blur")
print("Press 4 -> Canny Edge Detection")
print("Press 5 -> Binary Threshold")
print("Press 6 -> Original")
print("Press 'q' -> Exit")

while True:
    success, frame = cap.read()
    if not success:
        print("Failed to grab frame.")
        break

    # Apply effect based on selected mode key
    if mode == ord("1"):
        processed_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    elif mode == ord("2"):
        processed_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    elif mode == ord("3"):
        processed_frame = cv2.GaussianBlur(frame, (11, 11), 0)
    elif mode == ord("4"):
        # Apply Canny directly to grayscale for clean edge detection
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        processed_frame = cv2.Canny(gray, 50, 150)
    elif mode == ord("5"):
        # Threshold requires a 1-channel grayscale image first
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        _, processed_frame = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
    else:
        # Default / Mode 6: Original frame
        processed_frame = frame

    cv2.imshow("OpenCV Processing Lab", processed_frame)

    # Listen for key events continuously every 1 ms
    key = cv2.waitKey(1) & 0xFF

    if key in [ord("1"), ord("2"), ord("3"), ord("4"), ord("5"), ord("6")]:
        mode = key
    elif key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

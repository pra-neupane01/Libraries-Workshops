```python
import cv2
cap = cv2.VideoCapture(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Resources\video.mp4")
frame_count = 0

while cap.isOpened():
    success, frame = cap.read()
    if success == True:
        cv2.imshow("window",frame)
        filename = f"C:\\Users\\neupa\\Libraries-Workshops\\Open CV\\Tasks\\Extracted Image\\img_{frame_count}.png"
        cv2.imwrite(filename,frame)
        frame_count += 1

        if cv2.waitKey(25) & 0xFF == ord("p"):
            break
    else:
        break
cap.release()
cv2.destroyAllWindows()
```

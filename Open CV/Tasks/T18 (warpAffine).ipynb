```python
import cv2
import numpy as np
image = cv2.imread(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Resources\sharbhari.jpg")

image = cv2.resize(image, (400,400))

width, height = image.shape[:2]

matrix = np.float32([[0,1,100],[1,0,50]])

new_image = cv2.warpAffine(image, matrix, (width, height) )

cv2.imshow("window0",image)
cv2.imshow("window1",new_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

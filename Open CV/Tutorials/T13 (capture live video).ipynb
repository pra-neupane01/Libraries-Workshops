```python
import cv2
cap = cv2.VideoCapture(r"C:\Users\neupa\Libraries-Workshops\Open CV\Tasks\Resources\video.mp4")
while cap.isOpened():
    r,frame = cap.read()
    if r == True:
        frame = cv2.resize(frame,(500,500))
        cv2.imshow("window",frame)
        if cv2.waitKey(25) & 0xff == ord("p"):
            break
    else:
        break
cap.release()
cv2.destroyAllWindows()
```

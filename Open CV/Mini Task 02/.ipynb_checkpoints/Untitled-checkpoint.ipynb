```python
import cv2

input_path = r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Resources\video.mp4"
output_path = r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Resources\processed.mp4"

cap = cv2.VideoCapture(input_path)

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
output_path = r"C:\Users\neupa\Libraries-Workshops\Open CV\Tutorials\Resources\processed.mp4"

# Fixed parameter order: fourcc comes before fps
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

while cap.isOpened():
    r, frame = cap.read()
    if r:
        cv2.putText(
            frame,
            "Processed",
            (30, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1.0,
            (0, 255, 0),
            2
        )
        cv2.imshow("window", frame)
        
        out.write(frame)
        if cv2.waitKey(25) & 0xFF == ord("p"):
            break
    else:
        break

cap.release()
out.release()  # Essential to close and save the file properly
cv2.destroyAllWindows()
```
